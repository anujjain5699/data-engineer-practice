```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("Delete-Practice-Tables") \
    .getOrCreate()

# ── Delete Sales Table ─────────────────────────────────────────────
spark.sql("DROP TABLE IF EXISTS retailmart.sales_transactions")
print("✅ sales_transactions deleted")

# ── Delete Baseline Table ──────────────────────────────────────────
spark.sql("DROP TABLE IF EXISTS retailmart.optimization_baseline")
print("✅ optimization_baseline deleted")

# ── Verify Both Gone ───────────────────────────────────────────────
spark.sql("SHOW TABLES IN retailmart").show()
# Should show empty or no tables listed

```
